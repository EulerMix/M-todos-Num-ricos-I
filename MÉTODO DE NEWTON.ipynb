```python
# ==========================================================
# SOLUCIÓN DEL EJEMPLO 1: f(x) = cos(x) - x = 0
# a) Método de Punto Fijo | b) Método de Newton (Burden)
# ==========================================================

import numpy as np

tol = 1e-5
max_iter = 20

print("=" * 65)
print(" a) MÉTODO DE PUNTO FIJO: x = cos(x)")
print("=" * 65)
print(f"{'n':<5} {'p_n':<15} {'Error (|p_n - p_{n-1}|)':<25}")
print("-" * 45)

p0_fp = 0.5
g = lambda x: np.cos(x)
p = p0_fp

for i in range(1, max_iter + 1):
    p_next = g(p)
    error = abs(p_next - p)
    print(f"{i:<5} {p_next:<15.8f} {error:<25.8e}")
    if error < tol:
        print(f"\n=> Raíz aproximada (Punto Fijo): p = {p_next:.8f} en {i} iteraciones.\n")
        break
    p = p_next


print("=" * 65)
print(" b) MÉTODO DE NEWTON (Algoritmo 2.3 - Burden)")
print("=" * 65)
print(f"{'n':<5} {'p_n':<15} {'f(p_n)':<18} {'Error (|p_n - p_{n-1}|)':<25}")
print("-" * 65)

p0_newton = 0.5
f = lambda x: np.cos(x) - x
df = lambda x: -np.sin(x) - 1
p_n = p0_newton

for i in range(1, max_iter + 1):
    p_next = p_n - f(p_n) / df(p_n)
    error = abs(p_next - p_n)
    print(f"{i:<5} {p_next:<15.8f} {f(p_next):<18.8e} {error:<25.8e}")
    if error < tol:
        print(f"\n=> Raíz aproximada (Newton): p = {p_next:.8f} en {i} iteraciones.")
        break
    p_n = p_next
```
